# Objective

This code will separate your downloaded AF3 files based upon file type. Keep all files in one source directory, like your downloads folder, for it to work properly. 

```python
import os
import shutil
```

```python
# Define your source directory and the destination directories
source_dir = 'Path/To/Your/Downloaded/Files'
dest_dir_model = '/Path/To/Your/models'
dest_dir_full_data = '/Path/To/Your/full_data'
dest_dir_summary_confidences = '/Path/To/Your/summary_confidences'
dest_dir_job_request = '/Path/To/Your/job_requests'

# Create destination directories if they don't exist
for dest in [dest_dir_model, dest_dir_full_data, dest_dir_summary_confidences, dest_dir_job_request]:
    os.makedirs(dest, exist_ok=True)

# Walk through the directory tree
for root, dirs, files in os.walk(source_dir):
    for file in files:
        file_path = os.path.join(root, file)
        
        # Check if the file name contains one of the keywords and move accordingly
        if "model" in file:
            shutil.move(file_path, os.path.join(dest_dir_model, file))
            print(f"Moved {file} to {dest_dir_model}")

        elif "full_data" in file:
            shutil.move(file_path, os.path.join(dest_dir_full_data, file))
            print(f"Moved {file} to {dest_dir_full_data}")

        elif "summary_confidences" in file:
            shutil.move(file_path, os.path.join(dest_dir_summary_confidences, file))
            print(f"Moved {file} to {dest_dir_summary_confidences}")

        elif "job_request" in file:
            shutil.move(file_path, os.path.join(dest_dir_job_request, file))
            print(f"Moved {file} to {dest_dir_job_request}")

        else:
            print(f"File {file} did not match any known naming convention.")
```
